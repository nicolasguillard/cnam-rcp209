# Frame and Crop

```python
import os
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from torchvision import transforms as T
from torchvision.io import read_image
```

> Attention : utilisation d'un répertoire de données en dehors de l'environnement do notebook !

```python
#DataPath = "/Users/me/Documents/Work/Dev/_data/imagene_val_images"
DataPath = "/Users/me/Documents/Work/Dev/_data/imagenet-sample-images-master"
ImageFile = "n02128757_snow_leopard.JPEG"
filename = os.path.join(DataPath, ImageFile)
img1_t = read_image(filename)
print(img1_t.size(), img1_t.max().item())
img1 = T.functional.to_pil_image(img1_t)
display(img1)
```

# Frame

```python
receptive_field = ((50, 100), (75, 125))
```

```python
rf_min, rf_max = receptive_field
fig, ax = plt.subplots()
ax.imshow(img1)
rect = patches.Rectangle(
    rf_min[::-1], # (x,y)
    receptive_field[1][1] - receptive_field[0][1], # width
    receptive_field[1][0] - receptive_field[0][0], # height
    lw=1.5, edgecolor='black', fc='none', ls="-", joinstyle="round"
    )
ax.add_patch(rect)
plt.axis("off")
plt.show()
```

# Crop

```python
img2_t = img1_t[:, rf_min[0]:rf_max[0], rf_min[1]:rf_max[1]]
img2 = T.functional.to_pil_image(img2_t)
display(img2)
```
